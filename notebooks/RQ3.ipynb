```python
import json
from collections import defaultdict
import os

# 1. CONFIGURATION

# Update these paths to match your actual files
JSONL_FILE_PATH = "parsed_extracted_data.jsonl"  # Path to your JSONL file
OUTPUT_FILE = "analysis_results.json"  # Where to save results

# Make sure your ai_algorithms_dict is defined
# If it's in a separate file, import it here:

# 2. MAIN ANALYSIS FUNCTION


def analyze_papers(jsonl_file_path, algorithm_dict):
    """
    Analyze JSONL file containing papers and categorize their models.

    Args:
        jsonl_file_path: Path to JSONL file
        algorithm_dict: Dictionary mapping algorithms to categories/subcategories

    Returns:
        Dictionary containing all analysis results
    """

    print(f"Reading papers from: {jsonl_file_path}")
    print(f"Using algorithm dictionary with {len(algorithm_dict)} entries")

    # Read papers from JSONL file
    papers = []
    line_count = 0
    error_count = 0

    try:
        with open(jsonl_file_path, "r", encoding="utf-8") as f:
            for line_num, line in enumerate(f, 1):
                line_count += 1
                if line.strip():  # Skip empty lines
                    try:
                        paper = json.loads(line.strip())
                        papers.append(paper)
                    except json.JSONDecodeError as e:
                        error_count += 1
                        print(f"  Error parsing line {line_num}: {e}")
                        print(f"  Line content: {line[:100]}...")
    except FileNotFoundError:
        print(f"ERROR: File not found at {jsonl_file_path}")
        return None
    except Exception as e:
        print(f"ERROR reading file: {e}")
        return None

    print(f"\nSuccessfully loaded {len(papers)} papers")
    print(f"Total lines processed: {line_count}")
    if error_count > 0:
        print(f"Errors encountered: {error_count}")

    # Initialize data structures
    algorithm_counts = defaultdict(int)
    category_counts = defaultdict(int)
    subcategory_counts = defaultdict(int)
    algorithm_papers = defaultdict(set)
    category_papers = defaultdict(set)
    subcategory_papers = defaultdict(set)
    unknown_models = defaultdict(set)  # Track unknown models

    # Process each paper
    print("\n" + "=" * 70)
    print("PROCESSING PAPERS:")
    print("=" * 70)

    for paper in papers:
        paper_id = paper.get("ID", f"Unknown_{papers.index(paper)}")
        models = paper.get("Model", [])

        if not models:
            print(f"Paper {paper_id}: No models listed")
            continue

        # Track unique categories and subcategories for this paper
        paper_categories = set()
        paper_subcategories = set()
        paper_has_known_models = False

        print(f"\nPaper {paper_id}: {len(models)} model(s)")

        # Process each model in the paper
        for model in models:
            if model in algorithm_dict:
                paper_has_known_models = True
                algo_info = algorithm_dict[model]
                category = algo_info["category"]
                subcategory = algo_info["subcategory"]

                # Add to paper's unique sets
                paper_categories.add(category)
                paper_subcategories.add(subcategory)

                # Update algorithm counts
                algorithm_counts[model] += 1

                # Add paper to algorithm's set
                algorithm_papers[model].add(paper_id)

                print(f"  ✓ {model}")
                print(f"    → Category: {category}")
                print(f"    → Subcategory: {subcategory}")
            else:
                unknown_models[model].add(paper_id)
                print(f"  ✗ {model} (not in dictionary)")

        # Update category and subcategory counts for this paper
        if paper_has_known_models:
            for category in paper_categories:
                category_counts[category] += 1
                category_papers[category].add(paper_id)

            for subcategory in paper_subcategories:
                subcategory_counts[subcategory] += 1
                subcategory_papers[subcategory].add(paper_id)

    # 3. CALCULATE STATISTICS AND PREPARE OUTPUT

    # Prepare subcategory to algorithm mapping for top 5 analysis
    subcategory_algorithms = defaultdict(list)
    for algo, count in algorithm_counts.items():
        if algo in algorithm_dict:
            subcategory = algorithm_dict[algo]["subcategory"]
            subcategory_algorithms[subcategory].append((algo, count))

    # Sort each subcategory's algorithms by count
    for subcat in subcategory_algorithms:
        subcategory_algorithms[subcat].sort(key=lambda x: x[1], reverse=True)

    # Calculate statistics
    total_papers = len(papers)
    papers_with_known_models = sum(
        1
        for paper in papers
        if any(m in algorithm_dict for m in paper.get("Model", []))
    )
    total_algorithm_mentions = sum(len(paper.get("Model", [])) for paper in papers)

    # Prepare final output structure
    output_data = {
        "metadata": {
            "input_file": jsonl_file_path,
            "algorithm_dictionary_size": len(algorithm_dict),
        },
        "statistics": {
            "total_papers": total_papers,
            "papers_with_models": sum(1 for p in papers if p.get("Model")),
            "papers_with_known_models": papers_with_known_models,
            "total_algorithm_mentions": total_algorithm_mentions,
            "unique_algorithms_found": len(algorithm_counts),
            "categories_found": len(category_counts),
            "subcategories_found": len(subcategory_counts),
            "unknown_models_count": len(unknown_models),
        },
        "unknown_models": {
            model: {"paper_count": len(paper_ids), "papers": sorted(list(paper_ids))}
            for model, paper_ids in unknown_models.items()
        },
        "algorithm_counts": dict(
            sorted(algorithm_counts.items(), key=lambda x: x[1], reverse=True)
        ),
        "category_counts": dict(
            sorted(category_counts.items(), key=lambda x: x[1], reverse=True)
        ),
        "subcategory_counts": dict(
            sorted(subcategory_counts.items(), key=lambda x: x[1], reverse=True)
        ),
        "top_algorithms_by_subcategory": {
            subcat: algorithms[:5]
            for subcat, algorithms in subcategory_algorithms.items()
        },
        "papers_by_algorithm": {
            k: sorted(list(v)) for k, v in algorithm_papers.items()
        },
        "papers_by_category": {k: sorted(list(v)) for k, v in category_papers.items()},
        "papers_by_subcategory": {
            k: sorted(list(v)) for k, v in subcategory_papers.items()
        },
    }

    return output_data


# 4. DISPLAY RESULTS FUNCTION


def display_results(results):
    """Display analysis results in a readable format."""

    if not results:
        print("No results to display")
        return

    print("\n" + "=" * 70)
    print("ANALYSIS RESULTS SUMMARY")
    print("=" * 70)

    stats = results["statistics"]
    print(f"\n📊 STATISTICS:")
    print(f"   Total papers: {stats['total_papers']}")
    print(f"   Papers with models: {stats['papers_with_models']}")
    print(f"   Papers with known models: {stats['papers_with_known_models']}")
    print(f"   Total algorithm mentions: {stats['total_algorithm_mentions']}")
    print(f"   Unique algorithms found: {stats['unique_algorithms_found']}")
    print(f"   Categories found: {stats['categories_found']}")
    print(f"   Subcategories found: {stats['subcategories_found']}")
    print(f"   Unknown models: {stats['unknown_models_count']}")

    # Show unknown models if any
    if results["unknown_models"]:
        print(f"\n⚠️  UNKNOWN MODELS FOUND ({len(results['unknown_models'])}):")
        for model, info in results["unknown_models"].items():
            print(f"   '{model}' in {info['paper_count']} papers: {info['papers']}")

    print("\n" + "=" * 70)
    print("TOP 10 ALGORITHMS BY PAPER COUNT:")
    print("=" * 70)
    top_10 = list(results["algorithm_counts"].items())[:10]
    for i, (algo, count) in enumerate(top_10, 1):
        print(f"{i:2}. {algo:40} → {count:3} papers")

    print("\n" + "=" * 70)
    print("CATEGORY DISTRIBUTION:")
    print("=" * 70)
    for category, count in results["category_counts"].items():
        print(f"{category:50} → {count:3} papers")

    print("\n" + "=" * 70)
    print("SUBCATEGORY DISTRIBUTION:")
    print("=" * 70)
    for subcategory, count in results["subcategory_counts"].items():
        print(f"{subcategory:60} → {count:3} papers")

    print("\n" + "=" * 70)
    print("TOP ALGORITHMS BY SUBCATEGORY:")
    print("=" * 70)
    for subcat, top_algorithms in results["top_algorithms_by_subcategory"].items():
        print(f"\n{'-' * 40}")
        print(f"{subcat}:")
        if top_algorithms:
            for i, (algo, count) in enumerate(top_algorithms, 1):
                print(f"  {i}. {algo} ({count} papers)")
        else:
            print("  No algorithms found")

    # Show sample of paper distributions
    print("\n" + "=" * 70)
    print("SAMPLE PAPER DISTRIBUTIONS:")
    print("=" * 70)

    # Show first 3 algorithms and their papers
    print("\nPapers by algorithm (sample):")
    for i, (algo, papers_list) in enumerate(
        list(results["papers_by_algorithm"].items())[:3]
    ):
        print(
            f"  {algo}: {len(papers_list)} papers → {papers_list[:5]}{'...' if len(papers_list) > 5 else ''}"
        )

    # Show first 2 categories and their papers
    print("\nPapers by category (sample):")
    for i, (category, papers_list) in enumerate(
        list(results["papers_by_category"].items())[:2]
    ):
        print(f"  {category}: {len(papers_list)} papers")


# 5. SAVE RESULTS FUNCTION


def save_results(results, output_file):
    """Save analysis results to JSON file."""
    try:
        with open(output_file, "w", encoding="utf-8") as f:
            json.dump(results, f, indent=2, ensure_ascii=False)
        print(f"\n✅ Results saved to: {output_file}")
        print(f"   File size: {os.path.getsize(output_file):,} bytes")
        return True
    except Exception as e:
        print(f"\n❌ Error saving results: {e}")
        return False


# 6. MAIN EXECUTION


def main():
    """Main function to run the analysis."""

    print("=" * 70)
    print("PAPER ANALYSIS TOOL")
    print("=" * 70)

    # Check if files exist
    if not os.path.exists(JSONL_FILE_PATH):
        print(f"Error: JSONL file not found at {JSONL_FILE_PATH}")
        print("Please update JSONL_FILE_PATH in the code with your actual file path.")
        return

    # Run analysis
    results = analyze_papers(JSONL_FILE_PATH, ai_algorithms_dict)

    if results:
        # Display results
        display_results(results)

        # Save results
        if save_results(results, OUTPUT_FILE):
            print(f"\n📁 Output includes:")
            print(f"   - Statistics and counts")
            print(f"   - Unknown models found")
            print(f"   - Top algorithms by subcategory")
            print(f"   - Papers organized by algorithm, category, and subcategory")
            print(f"   - Complete mapping data")

    print("\n" + "=" * 70)
    print("ANALYSIS COMPLETE")
    print("=" * 70)


# 7. QUERY FUNCTIONS (for additional exploration)


def query_results(results, query_type, query_term):
    """Query the results for specific information."""

    if query_type == "algorithm":
        if query_term in results["papers_by_algorithm"]:
            papers = results["papers_by_algorithm"][query_term]
            count = results["algorithm_counts"].get(query_term, 0)
            print(f"\nAlgorithm: {query_term}")
            print(f"Papers using it: {count}")
            print(f"Paper IDs: {papers}")
        else:
            print(f"\nAlgorithm '{query_term}' not found in results")

    elif query_type == "category":
        if query_term in results["papers_by_category"]:
            papers = results["papers_by_category"][query_term]
            count = results["category_counts"].get(query_term, 0)
            print(f"\nCategory: {query_term}")
            print(f"Papers in this category: {count}")
            print(f"Paper IDs: {papers}")
        else:
            print(f"\nCategory '{query_term}' not found in results")

    elif query_type == "paper":
        # Search for a specific paper in all mappings
        print(f"\nSearching for paper: {query_term}")
        found_in = []

        for algo, papers in results["papers_by_algorithm"].items():
            if query_term in papers:
                found_in.append(f"Algorithm: {algo}")

        for category, papers in results["papers_by_category"].items():
            if query_term in papers:
                found_in.append(f"Category: {category}")

        if found_in:
            print(f"Paper {query_term} found in:")
            for item in found_in:
                print(f"  - {item}")
        else:
            print(f"Paper {query_term} not found in results")

    else:
        print(
            f"Unknown query type: {query_type}. Use 'algorithm', 'category', or 'paper'."
        )


# 8. RUN THE ANALYSIS

if __name__ == "__main__":
    # Run the main analysis
    main()

    # Example of how to query results after analysis
    # (Uncomment and modify as needed)
    """
    print("\n" + "=" * 70)
    print("EXAMPLE QUERIES:")
    print("=" * 70)
    
    # Load results for querying
    with open(OUTPUT_FILE, 'r') as f:
        results = json.load(f)
    
    # Query examples
    query_results(results, "algorithm", "SVM")
    query_results(results, "category", "Classical Machine Learning")
    query_results(results, "paper", "P1")
    """
```

```python
import json
from collections import defaultdict

JSONL_PATH = "parsed_extracted_data.jsonl"  # <-- your file

taxonomy = ai_algorithms_dict  # model -> {"category": "...", "subcategory": "..."}
task_map = TASK_RENAME_MAP  # normalize task names

# Optional: if your dataset uses older category names, normalize them here
CATEGORY_RENAME_MAP = {
    "Deep Learning": "Deep Learning (Task-Specific Models)",
    "Transformer-Based Models": "Transformer-Based Representation Models",
    "Large Language Models": "Transformer-Based Generative Models (LLMs)",
}


def norm_task(t: str) -> str:
    return task_map.get(t, t)


def norm_category(cat: str) -> str:
    return CATEGORY_RENAME_MAP.get(cat, cat)


# 1) Build: (category, subcategory) -> {tasks}, and -> {paperIDs}
cs_to_tasks = defaultdict(set)  # (cat, sub) -> set(tasks)
cs_to_papers = defaultdict(set)  # (cat, sub) -> set(paper IDs)

unknown_models = defaultdict(set)

with open(JSONL_PATH, "r", encoding="utf-8") as f:
    for line_num, line in enumerate(f, start=1):
        paper = json.loads(line)
        pid = paper.get("ID", f"UNKNOWN_ID_line_{line_num}")

        tasks = [norm_task(t) for t in paper.get("Task", [])]
        models = paper.get("Model", [])

        # collect (category, subcategory) pairs used by this paper
        used_pairs = set()
        for m in models:
            if m in taxonomy:
                cat = norm_category(taxonomy[m]["category"])
                sub = taxonomy[m]["subcategory"]
                used_pairs.add((cat, sub))
            else:
                unknown_models[m].add(pid)

        for (cat, sub) in used_pairs:
            cs_to_papers[(cat, sub)].add(pid)
            for t in tasks:
                cs_to_tasks[(cat, sub)].add(t)

# 2) For nicer LaTeX: group rows by category
cat_to_subs = defaultdict(list)
for (cat, sub) in cs_to_papers.keys():
    cat_to_subs[cat].append(sub)

for cat in cat_to_subs:
    cat_to_subs[cat] = sorted(set(cat_to_subs[cat]))

category_order = [
    "Classical Machine Learning",
    "Deep Learning (Task-Specific Models)",
    "Transformer-Based Representation Models",
    "Transformer-Based Generative Models (LLMs)",
]
# Append any extra categories not in the preferred order
for c in sorted(cat_to_subs.keys()):
    if c not in category_order:
        category_order.append(c)

# 3) LaTeX escaping helpers
def latex_escape(s: str) -> str:
    # Minimal escaping for common LaTeX troublemakers
    return (
        s.replace("&", r"\&")
        .replace("%", r"\%")
        .replace("_", r"\_")
        .replace("#", r"\#")
    )


def format_tasks(task_set):
    if not task_set:
        return r"\textit{None reported}"
    # sort for stability
    tasks = sorted(task_set)
    return "; ".join(latex_escape(t) for t in tasks)


def format_papers(paper_ids):
    if not paper_ids:
        return r"\textit{None}"
    ids = sorted(paper_ids)
    cites = ", ".join([rf"\citep{{{latex_escape(i)}}}" for i in ids])
    return cites


# 4) Build LaTeX longtable
lines = []
lines.append(r"% Auto-generated table: Category → Subcategory → Tasks → Papers")
lines.append(
    r"\begin{longtable}{p{0.23\linewidth} p{0.25\linewidth} p{0.27\linewidth} p{0.25\linewidth}}"
)
lines.append(
    r"\caption{Mapping between AI model categories/subcategories and the code review task categories addressed by the corresponding primary studies.}\label{tab:taxonomy_tasks}\\"
)
lines.append(r"\toprule")
lines.append(
    r"\textbf{Category} & \textbf{Subcategory} & \textbf{Task categories} & \textbf{Papers} \\"
)
lines.append(r"\midrule")
lines.append(r"\endfirsthead")

lines.append(r"\toprule")
lines.append(
    r"\textbf{Category} & \textbf{Subcategory} & \textbf{Task categories} & \textbf{Papers} \\"
)
lines.append(r"\midrule")
lines.append(r"\endhead")

lines.append(r"\midrule")
lines.append(r"\multicolumn{4}{r}{\footnotesize Continued on next page} \\")
lines.append(r"\endfoot")

lines.append(r"\bottomrule")
lines.append(r"\endlastfoot")

for cat in category_order:
    subs = cat_to_subs.get(cat, [])
    if not subs:
        continue
    for sub in subs:
        tasks_str = format_tasks(cs_to_tasks.get((cat, sub), set()))
        papers_str = format_papers(cs_to_papers.get((cat, sub), set()))
        lines.append(
            f"{latex_escape(cat)} & {latex_escape(sub)} & {tasks_str} & {papers_str} \\\\"
        )
    lines.append(r"\addlinespace")

lines.append(r"\end{longtable}")

out_tex = "\n".join(lines)

with open("taxonomy_tasks_table.tex", "w", encoding="utf-8") as f:
    f.write(out_tex)

print("Saved: taxonomy_tasks_table.tex")

# 5) Optional: log unknown models clearly
if unknown_models:
    print(
        "\nUnknown models not found in ai_algorithms_dict (model -> #papers, examples):"
    )
    for m, pids in sorted(unknown_models.items(), key=lambda x: (-len(x[1]), x[0])):
        ex = sorted(list(pids))[:8]
        print(f"- {m}: {len(pids)} papers (e.g., {ex})")
```

```python
import json
from collections import defaultdict

TASK_RENAME_MAP = {
    "Code refinement": "Code Refinement",
    "Code review data curation": "Code Review Data Curation",
    "Code review gamification": "Code Review Gamification",
    "Comment Update": "Comment Update",
    "Issue type prediction": "Issue Type Prediction",
    "Issue/bug detection": "Issue/Bug Detection",
    "Named Entity Recognition": "Named Entity Recognition",
    "Pull request completion time (effort estimation)": "Pull Request Effort Estimation",
    "Pull request prioritization": "Pull Request Prioritization",
    "Review comment evaluation": "Review Comment Evaluation",
    "Reviewer recommendation": "Reviewer Recommendation",
    "Sentiment analysis": "Sentiment Analysis",
    "code compliance assessment": "Code Compliance Assessment",
    "pull request assessment": "Pull Request Assessment",
    "review comment generation": "Review Comment Generation",
    "review comment recommendation": "Review Comment Recommendation",
}
TASK_TYPE_MAP = {
    "Pull Request Assessment": "Pre-Review",
    "Reviewer Recommendation": "Pre-Review",
    "Pull Request Prioritization": "Pre-Review",
    "Pull Request Effort Estimation": "Pre-Review",
    # Assisted
    "Issue/Bug Detection": "Assisted",
    "Issue Type Prediction": "Assisted",
    "Review Comment Recommendation": "Assisted",
    "Code Compliance Assessment": "Assisted",
    # Generative
    "Review Comment Generation": "Generative",
    "Code Refinement": "Generative",
    "Comment Update": "Generative",
    "Code Review Data Curation": "Generative",
    # Evaluation
    "Review Comment Evaluation": "Evaluation",
    "Sentiment Analysis": "Evaluation",
    "Named Entity Recognition": "Evaluation",
    "Code Review Gamification": "Evaluation",
}
```

```python
import json
from collections import defaultdict

# Inputs
JSONL_PATH = "parsed_extracted_data.jsonl"  # <-- your jsonl file (one paper per line)
taxonomy = ai_algorithms_dict  # model -> {"category": "...", "subcategory": "..."}
TASK_RENAME_MAP = TASK_RENAME_MAP  # your rename map (raw task -> canonical task)
TASK_TYPE_MAP = (
    TASK_TYPE_MAP  # your canonical task -> task type (Pre-Review/Assisted/...)
)

TASK_CAT_ORDER = ["Pre-Review", "Assisted", "Generative", "Evaluation"]

# Optional: preferred ordering of model categories in the table
MODEL_CAT_ORDER = [
    "Classical Machine Learning",
    "Deep Learning",
    "Transformer-Based Models",
    "Large Language Models",
]

# Helpers
def norm_task(t: str) -> str:
    return TASK_RENAME_MAP.get(t, t)


def latex_escape(s: str) -> str:
    # Minimal escaping for LaTeX special chars
    return (
        s.replace("&", r"\&")
        .replace("%", r"\%")
        .replace("_", r"\_")
        .replace("#", r"\#")
    )


def to_ps_bibkey(paper_id: str) -> str:
    """
    Convert paper IDs like 'P1' or 'p23' to BibTeX keys 'ps1', 'ps23'.
    If the ID doesn't start with 'P', we still prefix with 'ps'.
    """
    x = (paper_id or "").strip()
    if not x:
        return "ps_unknown"
    xl = x.lower()
    if xl.startswith("p") and xl[1:].isdigit():
        return "ps" + xl[1:]
    if xl.startswith("ps") and xl[2:].isdigit():
        return xl
    # fallback: prefix ps
    return "ps" + xl


def citep_group(ids) -> str:
    """
    Return ONE grouped citation command: \\citep{ps1,ps23,ps162}
    """
    ids = sorted(set(ids))
    if not ids:
        return "--"
    keys = [latex_escape(to_ps_bibkey(i)) for i in ids]
    return r"\citep{" + ",".join(keys) + "}"


def cat_sort_key(cat: str):
    return (MODEL_CAT_ORDER.index(cat) if cat in MODEL_CAT_ORDER else 999, cat)


# 1) Build cube: (modelCat, modelSub, taskCat) -> set(paperIDs)
cube = defaultdict(set)  # (cat, sub, taskCat) -> set(paper IDs)
unknown_models = defaultdict(set)  # model -> set(paper IDs)
unknown_tasks = defaultdict(set)  # task -> set(paper IDs)

with open(JSONL_PATH, "r", encoding="utf-8") as f:
    for line_num, line in enumerate(f, start=1):
        line = line.strip()
        if not line:
            continue

        paper = json.loads(line)
        pid = paper.get("ID", f"P_unknown_{line_num}")

        models = paper.get("Model", []) or []
        tasks_raw = paper.get("Task", []) or []

        # Normalize tasks and map to task categories
        task_cats_in_paper = set()
        for t in tasks_raw:
            t2 = norm_task(t)
            tc = TASK_TYPE_MAP.get(t2)
            if tc is None:
                unknown_tasks[t2].add(pid)
            else:
                task_cats_in_paper.add(tc)

        # Extract model (category, subcategory) pairs used by this paper
        pairs = set()
        for m in models:
            if m in taxonomy:
                cat = taxonomy[m]["category"]
                sub = taxonomy[m]["subcategory"]
                pairs.add((cat, sub))
            else:
                unknown_models[m].add(pid)

        # Fill cube
        for (cat, sub) in pairs:
            for tc in task_cats_in_paper:
                cube[(cat, sub, tc)].add(pid)

# 2) Rows (cat, sub) and ordering
# Group subcategories by model category
cat_to_subs = defaultdict(list)
for (cat, sub, _) in cube.keys():
    if sub not in cat_to_subs[cat]:
        cat_to_subs[cat].append(sub)

# Sort categories and subcategories
for cat in cat_to_subs:
    cat_to_subs[cat] = sorted(cat_to_subs[cat])

categories_ordered = sorted(cat_to_subs.keys(), key=cat_sort_key)


# 3) Generate LaTeX longtable
lines = []
lines.append(
    r"% Auto-generated table: model category/subcategory vs task categories (grouped under one header)"
)
lines.append(r"% Requires in LaTeX preamble: \usepackage{longtable,booktabs,natbib}")
lines.append(
    r"\begin{longtable}{p{0.22\linewidth} p{0.22\linewidth} p{0.14\linewidth} p{0.14\linewidth} p{0.14\linewidth} p{0.14\linewidth}}"
)
lines.append(
    r"\caption{AI model categories and subcategories mapped to grouped code review task categories. "
    r"Cells report the primary studies using each (model subcategory, task category) combination.}\label{tab:ai_models_task_categories}\\"
)
lines.append(r"\toprule")
lines.append(
    r"\textbf{Model category} & \textbf{Model subcategory} & \multicolumn{4}{c}{\textbf{Task categories (papers)}} \\"
)
lines.append(r"\cmidrule(lr){3-6}")
lines.append(
    r"& & \textbf{Pre-Review} & \textbf{Assisted} & \textbf{Generative} & \textbf{Evaluation} \\"
)
lines.append(r"\midrule")
lines.append(r"\endfirsthead")

lines.append(r"\toprule")
lines.append(
    r"\textbf{Model category} & \textbf{Model subcategory} & \multicolumn{4}{c}{\textbf{Task categories (papers)}} \\"
)
lines.append(r"\cmidrule(lr){3-6}")
lines.append(
    r"& & \textbf{Pre-Review} & \textbf{Assisted} & \textbf{Generative} & \textbf{Evaluation} \\"
)
lines.append(r"\midrule")
lines.append(r"\endhead")

lines.append(r"\bottomrule")
lines.append(r"\endlastfoot")

for cat in categories_ordered:
    subs = cat_to_subs[cat]
    n_subs = len(subs)

    for i, sub in enumerate(subs):
        vals = [citep_group(cube.get((cat, sub, tc), set())) for tc in TASK_CAT_ORDER]

        # Category cell only on the first row
        if i == 0:
            cat_cell = rf"\multirow{{{n_subs}}}{{*}}{{{latex_escape(cat)}}}"
        else:
            cat_cell = ""

        lines.append(
            f"{cat_cell} & {latex_escape(sub)} & "
            f"{vals[0]} & {vals[1]} & {vals[2]} & {vals[3]} \\\\"
        )

lines.append(r"\end{longtable}")

OUT_TEX = "ai_models_task_categories_table.tex"
with open(OUT_TEX, "w", encoding="utf-8") as f:
    f.write("\n".join(lines))

print(f"Saved: {OUT_TEX}")

# 4) Clear logs (as requested)
if unknown_models:
    print("\n[LOG] Models not found in ai_algorithms_dict:")
    for m, pids in sorted(unknown_models.items(), key=lambda x: (-len(x[1]), x[0])):
        ex = sorted(list(pids))[:8]
        print(f" - {m}: {len(pids)} papers (e.g., {ex})")

if unknown_tasks:
    print("\n[LOG] Tasks not found in TASK_TYPE_MAP (after TASK_RENAME_MAP):")
    for t, pids in sorted(unknown_tasks.items(), key=lambda x: (-len(x[1]), x[0])):
        ex = sorted(list(pids))[:8]
        print(f" - {t}: {len(pids)} papers (e.g., {ex})")
```
